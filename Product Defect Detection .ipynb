```python
import numpy as np
import pandas as pd
```

```python
def normalize(X):
    return (X - np.mean(X, axis=0)) / np.std(X, axis=0)
```

```python
def train_test_split(X, y, test_size=0.2):
    n = len(X)
    split_index = int(n * (1 - test_size))
    return X[:split_index], X[split_index:], y[:split_index], y[split_index:]
```

```python
def initialize_weights(n_features):
    weights = np.random.rand(n_features)
    bias = 0
    return weights, bias
```

```python
def predict(X, weights, bias):
    return np.dot(X, weights) + bias
```

```python
def compute_loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
def compute_gradients(X, y_true, y_pred):
    n = len(y_true)
    gradient_weights = -2 / n * np.dot(X.T, (y_true - y_pred))
    gradient_bias = -2 / n * np.sum(y_true - y_pred)
    return gradient_weights, gradient_bias
```

```python
def update_weights(weights, bias, gradient_weights, gradient_bias, learning_rate):
    weights -= learning_rate * gradient_weights
    bias -= learning_rate * gradient_bias
    return weights, bias
```

```python
def train_linear_regression(X, y, learning_rate=0.01, epochs=1000):
    n_features = X.shape[1]
    weights, bias = initialize_weights(n_features)
    losses = []

    for epoch in range(epochs):
        # Compute predictions
        y_pred = predict(X, weights, bias)
        loss = compute_loss(y, y_pred)
        losses.append(loss)
        gradient_weights, gradient_bias = compute_gradients(X, y, y_pred)
        weights, bias = update_weights(weights, bias, gradient_weights, gradient_bias, learning_rate)
        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss: {loss:.4f}")
    
    return weights, bias, losses
```

```python
def evaluate_model(X, y, weights, bias):
    y_pred = predict(X, weights, bias)
    loss = compute_loss(y, y_pred)
    return loss
```

```python
data = pd.read_csv('product_defect_data.csv')
data
```

```python
X = data[['SensorReadings', 'Temperature', 'Vibration', 'ProductionSpeed']].values
y = data['DefectRate'].values
```

```python
X = normalize(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
learning_rate = 0.01
epochs = 1000
weights, bias, losses = train_linear_regression(X_train, y_train, learning_rate, epochs)
```

```python
test_loss = evaluate_model(X_test, y_test, weights, bias)
print(f"Test Loss: {test_loss:.4f}")
```
